```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/interim/cleaned_day2.csv')
```

# Univariate Analysis

Numerical Univariate Analysis

```python
numerical_columns = [
    'Credit_Score',
    'Age',
    'Tenure',
    'Balance',
    'Num_Of_Products',
    'Estimated_Salary'
]

for col in numerical_columns:

    plt.figure(figsize=(8,4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.tight_layout()
    plt.savefig(f'../reports/figure/{col}_hist_kde.png')
                
    plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df['Credit_Score'])
plt.title('Boxplot of Credit Score')
plt.tight_layout()
plt.savefig('../reports/figure/creditscore_boxplot.png')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df['Estimated_Salary'])
plt.title('Boxplot of Esimated Salary')
plt.tight_layout()
plt.savefig('../reports/figure/salary_boxplot.png')

plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df['Age'])
plt.title('Boxplot of Age')
plt.tight_layout()

plt.savefig('../reports/figure/Age_boxplot.png')

plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df['Tenure'])
plt.title('Boxplot of Tenure')
plt.tight_layout()

plt.savefig('../reports/figure/Tenure_boxplot.png')

plt.show()
```

Categorical Univariate Analysis

```python
categorical_columns =[
    'Geography',
    'Gender',
    'Account_Type',
    'Has_Cr_Card',
    'Is_Active_Member',
    'Exited'
]

for col in categorical_columns:
    print("\n")
    print(df[col].value_counts())
```

```python
categorical_columns = [
    'Geography',
    'Gender',
    'Account_Type',
    'Has_Cr_Card',
    'Is_Active_Member',
    'Exited'
]

for col in categorical_columns:
    plt.figure(figsize=(8,4))
    sns.countplot(x=col, data=df)
    plt.title(f'Distribution of {col}')
    plt.tight_layout()

    plt.savefig(f'../reports/figure/{col}_barplot.png')
    plt.show()
```

correlation analysis

```python
correlation = df[numerical_columns].corr()

correlation
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(correlation, annot=True,cmap='coolwarm')
plt.title("Correlation Matrix")
plt.xticks(rotation=15)
plt.yticks(rotation=65)
plt.tight_layout()

plt.savefig("../reports/figure/correlation_heatmap.png")
plt.show()
```

Segment Analysis

```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[18,30,40,50,60,100],
    labels=['18-30','31-40','41-50','51-60','60+'])
```

```python
plt.figure(figsize=(8,6))

sns.countplot(x='Age_Group',hue='Exited',data=df)

plt.title('Age Group-wise Churn')
plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.tight_layout()

plt.savefig('../reports/figure/age_group_wise_churn.png')

plt.show()
```
