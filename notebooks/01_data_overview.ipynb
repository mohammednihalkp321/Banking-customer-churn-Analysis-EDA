# Dataset Description

- This dataset contains customer information from a bank.
- It includes customer demographics, account details, banking activity, and churn status.
- The dataset contains over 10,000 customer records and multiple numerical, categorical, and date features.
- Key attributes include Credit Score, Age, Balance, Number of Products, Account Type, and Customer Activity.
- The target variable is Exited, which indicates whether a customer has left the bank or remained active.
- The dataset also contains date columns that support time-based analysis and feature engineering.


# Business Understanding

- Customer churn is a major challenge in the banking industry.
- Understanding the factors that influence customer churn can help banks improve customer retention strategies.
- This analysis aims to identify patterns and characteristics associated with customers who leave the bank.
- The findings can help the bank take proactive actions to reduce churn and improve customer satisfaction.
- The project will explore customer demographics, financial behavior, account usage, and engagement levels to understand churn behavior.
